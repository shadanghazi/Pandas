```python
print(df.describe())
```

Country with highest population

```python
print(df[df['population'] == df['population'].max()]['country'])
```

Capital of the country with highest population

```python
print(df[df['population'] == df['population'].max()]['capital_city'])
```

Country with least population


```python
print(df[df['population'] == df['population'].min()]['country'])
```

Capital of the country with least population

```python
print(df[df['population'] == df['population'].min()]['capital_city'])
```

Top 5 countries with highest democracy


```python
df.sort_values(by="democracy_score", ascending=False, inplace=True)
print(df["country"].head())
```

Total regions

```python
print(df["region"].value_counts().count())
```

Total Countries in Eastern Europe

```python
print(df["region"].value_counts()["Eastern Europe"])
print(df[df["region"]== "Eastern Europe"]["country"])
```

political leader of the 2nd highest populated country

```python
print(df[df["population"]== df["population"].nlargest(2).iloc[1]]["political_leader"])
```

Number of countries whoes political leaders are unknown

```python
print(df[df["political_leader"].isna()]["country"].count())
```

countries which have Republic in their full name

```python
count= 0
def counting(text):
  global count
  if "republic" in text.lower():
    count += 1
  return text
df["country_long"].apply(counting)
print(count)
```

country in african region has highest population

```python
african_df= df[df["continent"]=="Africa"]
print(african_df)
print(african_df[african_df["population"]==african_df["population"].max()]["country"])
```
