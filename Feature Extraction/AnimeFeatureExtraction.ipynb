Extracting and saving Episodes to a new column


```python

def extract_episode(text):
  check = False
  data= ""
  for i in text:
    if i == ")":
        check = False
        return data
    if check == True:
        data = data + i
    if i == '(':
        check = True

print(df["Title"].apply(extract_episode))

```

Creating a new column "Episodes" and saving the Episode Numbers to it


```python

df["Episodes"]= df["Title"].apply(extract_episode)
print(df.head())
print(df.tail())
```

Converting the Episodes Numbers to integers

```python
df["Episodes"]= df["Episodes"].str.replace(" eps","")
df["Episodes"]= df["Episodes"].astype(int)

print(df.head())
print(df.tail())
```

Extracting Timestamp from the title

```python
def extract_time(text):
  check = False
  data=""
  for i in range(len(text)):
    if text[i]== ")":
      for j in range(i+1,i+20):
        data+=text[j]
      return data

df["Total Time"]= df["Title"].apply(extract_time)
print(df["Total Time"])
```

Converting Timestamp to Integers

```python
from dateutil.relativedelta import relativedelta
from datetime import datetime

def calculate_total_months(period):
    try:
        start_str, end_str = period.split(' - ')
        start_date = datetime.strptime(start_str, '%b %Y')
        end_date = datetime.strptime(end_str, '%b %Y')
        r = relativedelta(end_date, start_date)
        return r.years * 12 + r.months + 1
    except:
        return None

df['Months'] = df['Total Time'].apply(calculate_total_months)
print(df["Months"])
```
